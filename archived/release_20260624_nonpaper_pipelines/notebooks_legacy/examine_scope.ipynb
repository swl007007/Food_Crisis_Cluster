```python
import pandas as pd
import numpy as np
import polars as pl




DATA_PATH  = r"C:\Users\swl00\IFPRI Dropbox\Weilun Shi\Google fund\Analysis\1.Source Data\FEWSNET_IPC_train_lag_forecast_v06252025.csv"

# read data using polars
df = pl.read_csv(DATA_PATH)
```

```python
# keep only date fews_ipc_crisis, FEWSNET_admin_code
df = df.select([
    "date",
    "fews_ipc_crisis",
    "FEWSNET_admin_code"
])
```

```python
# convert to pandas for easier manipulation
df = df.to_pandas()
```

```python
# convert date to datetime
df['date'] = pd.to_datetime(df['date'])
```

```python
# keep only rows where fews_ipc_crisis is not null
df = df[df['fews_ipc_crisis'].notnull()]
```

```python
#group by FEWSNET_admin_code, calculate the months difference of each date from the previous date
df['months_diff'] = df.groupby('FEWSNET_admin_code')['date'].diff().dt.days // 30
```

```python
# value counts of months_diff
df['months_diff'].value_counts().sort_index()
```
