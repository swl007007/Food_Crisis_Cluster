```python
# import data

import polars as pl
```

```python
DATA_PATH = r"C:\Users\swl00\IFPRI Dropbox\Weilun Shi\Google fund\Analysis\1.Source Data\FEWSNET_final_v06152025_with_zscores.csv"
```

```python
# import csv
df = pl.read_csv(DATA_PATH)
```

```python
# remove fews_ipc_crisis, then regenerate fews_ipc_crisis based on fews_ipc, if fews_ipc >= 3 then fews_ipc_crisis =1 if fews_ipc <3 then fews_ipc_crisis=0
df = df.with_columns(
    pl.when(pl.col("fews_ipc") >= 3)
    .then(1)
    .alias("fews_ipc_crisis")
)
```

```python
# edit fews_ipc_crisis to be 0 if fews_ipc<3 using polars, do not modify other rows
df = df.with_columns(
    pl.when(pl.col("fews_ipc") < 3)
    .then(0)
    .otherwise(pl.col("fews_ipc_crisis"))
    .alias("fews_ipc_crisis")
)

```

```python
# rename admin_code to FEWSNET_admin_code
df = df.rename({"admin_code": "FEWSNET_admin_code"})
```

```python
#assemble date from year and month
df = df.with_columns(
    pl.concat_str([pl.col("year").cast(pl.Utf8), pl.col("month").cast(pl.Utf8).str.zfill(2)], separator="-").alias("date")
)
```
